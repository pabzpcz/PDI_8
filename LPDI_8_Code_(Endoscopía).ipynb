#Parte 1

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('/content/Endoscopia.jpeg', cv2.IMREAD_GRAYSCALE)

plt.imshow(img, cmap='gray')
plt.axis('off')
plt.show()
```

##Filtros de Suavizado

###Filtro de Media

```python
img_media3 = cv2.blur(img, (3, 3))    # Kernel 3x3
img_media5 = cv2.blur(img, (5, 5))    # Kernel 5x5
img_media11 = cv2.blur(img, (11, 11)) # Kernel 11x11

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img_media3, cmap='gray')
plt.title('Media 3x3')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(img_media5, cmap='gray')
plt.title('Media 5x5')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(img_media11, cmap='gray')
plt.title('Media 11x11')
plt.axis('off')

plt.tight_layout()
plt.show()
```

###Filtro Gaussiano

```python
img_gauss3 = cv2.GaussianBlur(img, (3, 3), 0)    # 3x3
img_gauss5 = cv2.GaussianBlur(img, (5, 5), 0)    # 5x5
img_gauss11 = cv2.GaussianBlur(img, (11, 11), 0) # 11x11

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img_gauss3, cmap='gray')
plt.title('Gaussiano 3x3')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(img_gauss5, cmap='gray')
plt.title('Gaussiano 5x5')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(img_gauss11, cmap='gray')
plt.title('Gaussiano 11x11')
plt.axis('off')

plt.tight_layout()
plt.show()
```

###Filtro de Mediana

```python
img_mediana3 = cv2.medianBlur(img, 3)   # 3x3
img_mediana5 = cv2.medianBlur(img, 5)   # 5x5
img_mediana11 = cv2.medianBlur(img, 11) # 11x11

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img_mediana3, cmap='gray')
plt.title('Mediana 3x3')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(img_mediana5, cmap='gray')
plt.title('Mediana 5x5')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(img_mediana11, cmap='gray')
plt.title('Mediana 11x11')
plt.axis('off')

plt.tight_layout()
plt.show()
```

##Filtros de Detección de Bordes

###Valores Normales

```python
mask_sobel_x   = np.array([[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]])
mask_sobel_y   = np.array([[-1, -2, -1], [0, 0, 0], [1, 2, 1]])
mask_laplacian = np.array([[0, 1, 0], [1, -4, 1], [0, 1, 0]])
mask_kirsch    = np.array([[5, 5, 5], [-3, 0, -3], [-3, -3, -3]])

img_sobel_x   = np.abs(cv2.filter2D(img, cv2.CV_64F, mask_sobel_x))
img_sobel_y   = np.abs(cv2.filter2D(img, cv2.CV_64F, mask_sobel_y))
img_laplacian = np.abs(cv2.filter2D(img, cv2.CV_64F, mask_laplacian))
img_kirsch    = np.abs(cv2.filter2D(img, cv2.CV_64F, mask_kirsch))

plt.figure(figsize=(18, 5))

plt.subplot(1, 4, 1)
plt.imshow(img_sobel_x, cmap='gray')
plt.title('Sobel X Normales')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(img_sobel_y, cmap='gray')
plt.title('Sobel Y Normales')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(img_laplacian, cmap='gray')
plt.title('Laplaciano Normales')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(img_kirsch, cmap='gray')
plt.title('Kirsch Normales')
plt.axis('off')

plt.tight_layout()
plt.show()
```

###Valores Modificados

```python
mask_sobel_x_mod   = np.array([[-2, 0, 2], [-5, 0, 5], [-2, 0, 2]])
mask_sobel_y_mod   = np.array([[-2, -5, -2], [0, 0, 0], [2, 5, 2]])
mask_laplacian_mod = np.array([[1, 1, 1], [1, -8, 1], [1, 1, 1]])
mask_kirsch_mod    = np.array([[8, 8, 8], [-4, 0, -4], [-4, -4, -4]])

img_sobel_x_mod   = np.abs(cv2.filter2D(img, cv2.CV_64F, mask_sobel_x_mod))
img_sobel_y_mod   = np.abs(cv2.filter2D(img, cv2.CV_64F, mask_sobel_y_mod))
img_laplacian_mod = np.abs(cv2.filter2D(img, cv2.CV_64F, mask_laplacian_mod))
img_kirsch_mod    = np.abs(cv2.filter2D(img, cv2.CV_64F, mask_kirsch_mod))

plt.figure(figsize=(18, 5))

plt.subplot(1, 4, 1)
plt.imshow(img_sobel_x_mod, cmap='gray')
plt.title('Sobel X Modificado')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(img_sobel_y_mod, cmap='gray')
plt.title('Sobel Y Modificado')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(img_laplacian_mod, cmap='gray')
plt.title('Laplaciano Modificado')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(img_kirsch_mod, cmap='gray')
plt.title('Kirsch Modificado')
plt.axis('off')

plt.tight_layout()
plt.show()
```

##Máscara

```python
mask_realce = np.array([[-1, -1, -1], [-1,  9, -1], [-1, -1, -1]])
img_realce = cv2.filter2D(img, cv2.CV_64F, mask_realce)

img_realce = np.clip(img_realce, 0, 255)
img_realce = np.uint8(img_realce)

plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Radiografía Original')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_realce, cmap='gray')
plt.title('Radiografía con Filtro de Realce')
plt.axis('off')

plt.tight_layout()
plt.show()
```

#Parte 2

##Dominio de la Frecuencia con la Transformada de Fourier

```python
f_transformada = np.fft.fft2(img)
f_centrada = np.fft.fftshift(f_transformada)

espectro_magnitud = 20 * np.log(np.abs(f_centrada) + 1)

plt.figure(figsize=(12, 6))

# Dominio Espacial
plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Dominio Espacial (Radiografía)')
plt.axis('off')

# Dominio de la Frecuencia
plt.subplot(1, 2, 2)
plt.imshow(espectro_magnitud, cmap='gray')
plt.title('Dominio de la Frecuencia (Espectro)')
plt.axis('off')

plt.tight_layout()
plt.show()
```

##Filtros Pasa Bajas

```python
filas, columnas = img.shape
centro_fil, centro_col = filas // 2, columnas // 2

y, x = np.ogrid[:filas, :columnas]
distancias = np.sqrt((x - centro_col)**2 + (y - centro_fil)**2)

mask_lp20 = np.where(distancias <= 20, 1, 0)
mask_lp50 = np.where(distancias <= 50, 1, 0)
mask_lp80 = np.where(distancias <= 80, 1, 0)

f_filtrada_20 = f_centrada * mask_lp20
f_filtrada_50 = f_centrada * mask_lp50
f_filtrada_80 = f_centrada * mask_lp80

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(mask_lp20, cmap='gray')
plt.title('Pasa Bajas - Radio = 20')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(mask_lp50, cmap='gray')
plt.title('Pasa Bajas - Radio = 50')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(mask_lp80, cmap='gray')
plt.title('Pasa Bajas - Radio = 80')
plt.axis('off')

plt.tight_layout()
plt.show()
```

##Filtros Pasa Altas

```python
filas, columnas = img.shape
centro_fil, centro_col = filas // 2, columnas // 2

y, x = np.ogrid[:filas, :columnas]
distancias = np.sqrt((x - centro_col)**2 + (y - centro_fil)**2)

# Máscara circular
mask_hp_circ = np.where(distancias > 30, 1, 0)
f_filtrada_circ = f_centrada * mask_hp_circ

# Máscara en cruz
mask_hp_cruz = np.zeros((filas, columnas))
mask_hp_cruz[centro_fil-10:centro_fil+10, :] = 1
mask_hp_cruz[:, centro_col-10:centro_col+10] = 1
mask_hp_cruz[centro_fil-30:centro_fil+30, centro_col-30:centro_col+30] = 0
f_filtrada_cruz = f_centrada * mask_hp_cruz

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(mask_hp_circ, cmap='gray')
plt.title('Pasa Altas - Círculo')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(mask_hp_cruz, cmap='gray')
plt.title('Pasa Altas - Cruz')
plt.axis('off')

plt.tight_layout()
plt.show()
```

##Tranformada Inversa

###Filtros Pasa Bajas (Espacio)

```python
img_lp20 = np.abs(np.fft.ifft2(np.fft.ifftshift(f_filtrada_20)))
img_lp50 = np.abs(np.fft.ifft2(np.fft.ifftshift(f_filtrada_50)))
img_lp80 = np.abs(np.fft.ifft2(np.fft.ifftshift(f_filtrada_80)))

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img_lp20, cmap='gray')
plt.title('Pasa Bajas (Espacio) - Radio = 20')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(img_lp50, cmap='gray')
plt.title('Pasa Bajas (Espacio) - Radio = 50')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(img_lp80, cmap='gray')
plt.title('Pasa Bajas (Espacio) - Radio = 80')
plt.axis('off')

plt.tight_layout()
plt.show()
```

###Filtros Pasa ALtas (Espacio)

```python
img_hp_circ = np.abs(np.fft.ifft2(np.fft.ifftshift(f_filtrada_circ)))
img_hp_cruz = np.abs(np.fft.ifft2(np.fft.ifftshift(f_filtrada_cruz)))

img_hp_circ_visible = cv2.normalize(img_hp_circ, None, 0, 255, cv2.NORM_MINMAX)
img_hp_cruz_visible = cv2.normalize(img_hp_cruz, None, 0, 255, cv2.NORM_MINMAX)

plt.figure(figsize=(12, 5))

# Círculo
plt.subplot(1, 2, 1)
plt.imshow(img_hp_circ_visible, cmap='gray')
plt.title('Pasa Altas (Espacio) - Círculo')
plt.axis('off')

# Cruz
plt.subplot(1, 2, 2)
plt.imshow(img_hp_cruz_visible, cmap='gray')
plt.title('Pasa Altas (Espacio) - Cruz')
plt.axis('off')

plt.tight_layout()
plt.show()
```
